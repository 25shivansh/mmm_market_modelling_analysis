# Module 4: Constructing the Knowledge Base for Multi-RAG

## 1. What is Retrieval-Augmented Generation (RAG)?
Retrieval-Augmented Generation (RAG) is an AI architecture that enhances Large Language Models (LLMs) by dynamically connecting them to external, proprietary databases. Instead of relying solely on general pre-trained knowledge, a RAG system first retrieves relevant facts from a custom knowledge base and injects that context into the LLM's prompt. This enables the model to generate highly accurate, domain-specific, and verifiable answers.

## 2. Why do Large Language Models need external knowledge?
While LLMs are powerful reasoning engines, their training data is static, generalized, and public. Without external knowledge, they suffer from three critical business limitations:
- **Hallucinations:** They may confidently invent plausible but factually incorrect information.
- **Lack of Proprietary Context:** They cannot access internal company data, such as private marketing strategies, specific customer sentiment, or proprietary forecasts.
- **Outdated Knowledge:** Their knowledge cuts off at the time of their last training, making them blind to recent trends or newly generated reports.

Supplying external knowledge grounds the LLM in factual, real-time, organization-specific reality, ensuring reliable and strategic decision-making.

## 4. What is a Knowledge Base?
In the context of modern generative AI, a Knowledge Base is typically a **Vector Database**. Unlike traditional relational databases (SQL) that rely on exact keyword matches, a vector database stores documents that have been converted into mathematical representations called *embeddings*. This allows the system to perform semantic search—retrieving information based on the actual meaning and context of a user's query.
